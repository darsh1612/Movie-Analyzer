```python
## genre is important column, id is important column , keyword is important column, title is important column
# overview is important column, cast is important column, crew is important column, 

importantmovies = newdata[['movie_id','title','genres','keywords','cast','crew','overview']]
importantmovies.head()
```

```python
importantmovies.duplicated().sum()
```

```python
changemovies = importantmovies
changemovies.iloc[0].genres
```

```python
import ast 
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
changemovies['genres'] = changemovies['genres'].apply(convert)
changemovies.head()

```

```python
changemovies['keywords'] = changemovies['keywords'].apply(convert)
changemovies.head()
```

```python
changemovies['cast'][0]
def convert_cast(obj):
    l = []
    count = 0
    for i in ast.literal_eval(obj):
        if count != 3:
            l.append(i['name'])
            count += 1
        else:
            break
    return l
```

```python
changemovies['cast'] = changemovies['cast'].apply(convert_cast)
changemovies.head()
```

```python

## job is director
def get_director(crew):
    l = []
    for person in ast.literal_eval(crew):
        if person['job'] == 'Director':
            l.append(person['name'])
            return l
    return None
```

```python
changemovies['crew'] = changemovies['crew'].apply(get_director)
changemovies.head()
```

```python
changemovies['overview'] = changemovies['overview'].fillna('').apply(lambda x: x.split())
changemovies.head()
```

```python
def remove_spaces(x):
    if isinstance(x, list):
        return [i.replace(" ", "") for i in x]
    else:
        return []

changemovies['genres']   = changemovies['genres'].apply(remove_spaces)
changemovies['crew']     = changemovies['crew'].apply(remove_spaces)
changemovies['keywords'] = changemovies['keywords'].apply(remove_spaces)
changemovies['cast']     = changemovies['cast'].apply(remove_spaces)
changemovies.head()
```

```python
changemovies['tags'] = changemovies['overview'] + changemovies['genres'] + changemovies['keywords'] + changemovies['cast'] + changemovies['crew'] 
changemovies = changemovies.drop(columns=['genres', 'keywords', 'cast', 'crew', 'overview'])
changemovies['tags'][0]


```

```python
changemovies.head()
```

```python
## remove empty elemnts from tags ,, and lower eveyrhting
def clean_tags(x):
    return [i.lower() for i in x if i.strip() != '']
```

```python
changemovies['tags'] = changemovies['tags'].apply(clean_tags)
```

## NOW DO VECTORIZATION We have done preprocessing


```python
##now convert our movie tags to a vector using word2vec or bag of words 

import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
## Q&A,chatbots,text summarization
import nltk
# Download required NLTK data for lemmatization
nltk.download('wordnet')
nltk.download('omw-1.4')  # Open Multilingual Wordnet for better lemmatization
from nltk.stem import WordNetLemmatizer
from nltk.stem import PorterStemmer

ps = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
```

```python
def nlp_clean(text):
    words = text.split()

    cleaned = []
    for w in words:
        w = w.lower()
        if w not in stop_words:
            w = ps.lemmatize(w)     # running → run
            w = stemmer.stem(w)            # running → run
            cleaned.append(w)

    return " ".join(cleaned)
```

```python
# Convert list of tokens into a sentence
changemovies['tags'] = changemovies['tags'].apply(lambda x: " ".join(x))

# Now apply lemmatization + stemming
changemovies['tags'] = changemovies['tags'].apply(nlp_clean)


```

```python
changemovies['tags'][0]
```

```python
##apply vectorization

from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000)
vectors = cv.fit_transform(changemovies['tags']).toarray()
```

```python
vectors.shape  ## 4809 movies 5000 most used words made an vector for it 
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

```python
similarity.shape 
```

```python
similarity[0] ## similarity score of first movie with other movies 
```

```python
def recommend_movies(movie_name, num_recommendations=5):
    movie_index = changemovies[changemovies['title'] == movie_name].index[0]

    distances = similarity[movie_index]

    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:num_recommendations+1]

    recommended_movies = []
    for i in movies_list:
        recommended_movies.append(changemovies.iloc[i[0]].title)

    return recommended_movies
```

```python
recommend_movies("Batman")
```

```python
import pickle

# Keep only what the recommender needs
new_df = changemovies[['movie_id', 'title', 'tags']]

# Save movies dataframe
pickle.dump(new_df, open('movies.pkl', 'wb'))

# Save similarity matrix (the ML model)
pickle.dump(similarity, open('similarity.pkl', 'wb'))

print("movies.pkl and similarity.pkl created successfully!")
```
